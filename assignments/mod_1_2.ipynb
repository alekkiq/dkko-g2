```python
import numpy as np
import pandas as pd
from scipy.io import arff
import matplotlib.pyplot as plt
```

### Loading data and extracting only necessary colums

- age   	Age
- bp	    Blood Pressure
- sg	    Specific Gravity
- al	    Albumin
- su	    Sugar
- bgr	    Blood Glucose Random
- bu	    Blood Urea
- sod	    Sodium
- pot	    Potassium
- hemo	    Hemoglobin
- pcv	    Packed Cell Volume
- wbcc	    White Blood Cell Count
- rbcc	    Red Blood Cell Count
- class 	Class

```python
data = arff.loadarff('../datasets/ckd/chronic_kidney_disease_full.arff')
df = pd.DataFrame(data[0])
df = df[["age", "bp", "sg", "al", "su", "bgr", "bu", "sod", "pot", "hemo", "pcv", "wbcc", "rbcc", "class"]]
df.head(10)
```

### Converting Columns To Show Desired Values

```python
# Converting the "hemo" column to g/l from g/dl by dividing by 10.
df["hemo"] = df["hemo"] / 10
df.head(10)
```

```python
#show all types of class column
df["class"].unique()

#change b"ckd" = a, b"notckd" = c
df["class"] = df["class"].replace({b"ckd": "a", b"notckd": "c"})

#checking the first 10 rows of the dataframe after the changes
df.head(10)
```

### Cleaning Missing Values


```python
# Count missing values per row
missing_values_count = df.isna().sum(axis=1)

# Dropping rows with 3 or more missing values
df_cleaned = df[missing_values_count < 3]

df_cleaned.head(10)

# show the number or rows left after dropping rows with 3 or more missing values
print(f"Number of rows left after dropping rows with 3 or more missing values: {len(df_cleaned)}")
```

### Splitting Dataframes
Splitting the dataframes to seperate the affected and control individuals.


```python
df_affected_group = df_cleaned[df_cleaned["class"] == "a"]
df_control_group = df_cleaned[df_cleaned["class"] == "c"]

display(df_affected_group.head(10))
print(f"Number of rows in the affected group: {len(df_affected_group)}")

display(df_control_group.head(10))
print(f"Number of rows in the control group: {len(df_control_group)}")
```

### Statistics

```python
print("Descriptive statistics for the affected group:")
display(df_affected_group.describe(include='all'))

print("Descriptive statistics for the control group:")
display(df_control_group.describe(include='all'))
```

There are some values that are physically impossible to have, for example potassium in the affected group is recorded at maximum value of 47, which is physically impossible. The minimum value of sodium in the affected group is 4.5, which is way too low. These are certainly data errors, a human can't survive with those values. There are some more values that are slightly outside the norms like bloodpressure, but sodium and potassium were the most dramatic outliers.

We would handle them, by dropping the whole rows that have dramatic statistical errors. 

### Correlation Matrices



```python
# define the correlation matrices
df_cleaned_corr = df_cleaned.corr(numeric_only=True)
df_affected_group_corr = df_affected_group.corr(numeric_only=True)
df_control_group_corr = df_control_group.corr(numeric_only=True)
```

```python
def plot_correlation_matrix(corr_matrix, title):
    plt.figure(figsize=(10, 10))
    plt.imshow(corr_matrix, cmap='coolwarm', interpolation='none')
    plt.colorbar()
    plt.xticks(range(len(corr_matrix)), corr_matrix.columns, rotation=90)
    plt.yticks(range(len(corr_matrix)), corr_matrix.columns)
    plt.title(title)
    plt.show()
```

```python
plot_correlation_matrix(df_cleaned_corr, "Combined") # the original dataset
```

In the heatmap for the original combined dataset we see that there are some strong correlations:

- Hemoglobin, Packed Cell Volume and Red Blood Cell Count are all strongly positively correlated to each other, which makes sense since they are all related to red blood cells
- Hemoglobin, Packed Cell Volume and Red Blood Cell Count are strongly negatively correlated to Blood Urea. This could be because as kidney function declines, the body produces less red blood cells, leading to lower hemoglobin, packed cell volume and red blood cell count, while at the same time blood urea levels increase due to reduced kidney function

```python
plot_correlation_matrix(df_affected_group_corr, "Affected Group") # the affected group
```

The heatmap for the affected group shows lots of similar correlations to the combined dataset

```python
plot_correlation_matrix(df_control_group_corr, "Control Group") # the control group
```

The control group heatmap shows that there are no strong positive correlations between any features, but there are some strong negative correlations between Hemoglobin, Packed Cell Volume and Red Blood Cell Count with Blood Urea. This could be because in healthy individuals, the body produces more red blood cells, leading to higher hemoglobin, packed cell volume and red blood cell count, while at the same time blood urea levels decrease due to better kidney function.
