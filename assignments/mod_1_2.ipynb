```python
import numpy as np
import pandas as pd
from scipy.io import arff
```

### Loading data and extracting only necessary colums

- age   	Age
- bp	    Blood Pressure
- sg	    Specific Gravity
- al	    Albumin
- su	    Sugar
- bgr	    Blood Glucose Random
- bu	    Blood Urea
- sod	    Sodium
- pot	    Potassium
- hemo	    Hemoglobin
- pcv	    Packed Cell Volume
- wbcc	    White Blood Cell Count
- rbcc	    Red Blood Cell Count
- class 	Class

```python
data = arff.loadarff('../datasets/ckd/chronic_kidney_disease_full.arff')
df = pd.DataFrame(data[0])
df = df[["age", "bp", "sg", "al", "su", "bgr", "bu", "sod", "pot", "hemo", "pcv", "wbcc", "rbcc", "class"]]
df.head(10)
```

### Converting Columns To Show Desired Values

```python
# Converting the "hemo" column to g/l from g/dl by dividing by 10.
df["hemo"] = df["hemo"] / 10
df.head(10)
```

```python
#show all types of class column
df["class"].unique()

#change b"ckd" = a, b"notckd" = c
df["class"] = df["class"].replace({b"ckd": "a", b"notckd": "c"})

#checking the first 10 rows of the dataframe after the changes
df.head(10)
```

### Cleaning Missing Values


```python
# Count missing values per row
missing_values_count = df.isna().sum(axis=1)

# Dropping rows with 3 or more missing values
df_cleaned = df[missing_values_count < 3]

df_cleaned.head(10)

# show the number or rows left after dropping rows with 3 or more missing values
print(f"Number of rows left after dropping rows with 3 or more missing values: {len(df_cleaned)}")
```

### Splitting Dataframes
Splitting the dataframes to seperate the affected and control individuals.


```python
df_affected_group = df_cleaned[df_cleaned["class"] == "a"]
df_control_group = df_cleaned[df_cleaned["class"] == "c"]

display(df_affected_group.head(10))
print(f"Number of rows in the affected group: {len(df_affected_group)}")

display(df_control_group.head(10))
print(f"Number of rows in the control group: {len(df_control_group)}")
```

### Statistics

```python
print("Descriptive statistics for the affected group:")
display(df_affected_group.describe(include='all'))

print("Descriptive statistics for the control group:")
display(df_control_group.describe(include='all'))
```

There are some values that are physically impossible to have, for example potassium in the affected group is recorded at maximum value of 47, which is physically impossible. The minimum value of sodium in the affected group is 4.5, which is way too low. These are certainly data errors, a human can't survive with those values. There are some more values that are slightly outside the norms like bloodpressure, but sodium and potassium were the most dramatic outliers.

We would handle them, by dropping the whole rows that have dramatic statistical errors.
